```python
import pandas as pd
```

```python
df = pd.read_csv('./data/final.csv', index_col=0)
df.head(2)
```

```python
df.columns
```

```python
df = df[['y', 'msg_length', 'max_discount', 'emoji_count', 'time_pressure',
       'SRCH_EFRT', 'CAT_CNT',
       'AVG_DEPTH', 'day_off']]
```

```python
df.head(2)
```

```python
df_copy = df.copy()


def remove_outliers(df, columns):
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = 0
        upper_bound = Q3 + 1.5 * IQR
        df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]
    return df

df = remove_outliers(df, ["SRCH_EFRT", "CAT_CNT", "AVG_DEPTH"])
```

```python
import statsmodels.api as sm

# 특징 변수와 종속 변수 분리
X = df_copy.drop(columns=['y'])  # 'y'는 종속 변수 열 이름입니다.
y = df_copy['y']

# 상수항 추가
X = sm.add_constant(X)


# 로지스틱 회귀 모델 생성 및 학습
model = sm.Logit(y, X)
result = model.fit()

result.summary()
```
