```python
import pandas as pd
```

### 메시지 전처리

```python
msg = pd.read_csv('./raw/CUSTOMER_MSG.csv')
print(msg.shape)
msg.head(2)
```

```python
msg.iloc[0]['JOUR_NM'].split('_')[0]
```

```python
def jour(text):
    return text.split('_')[0]

msg['JOUR'] = msg['JOUR_NM'].apply(jour)
msg.head(2)
```

```python
msg_dtl = pd.read_csv('./raw/11월_친구톡utm.csv')
msg_dtl['발송시간'] = pd.to_datetime(msg_dtl['발송시간'])
msg_dtl.head(2)
```

```python
msg_merge = pd.merge(msg, msg_dtl, left_on='JOUR', right_on='저니', how='left')
msg_merge['본문'] = msg_merge['본문'].astype(str)
msg_merge = msg_merge[~msg_merge['본문'].str.contains('\[안내\]')]
print(msg_merge.shape)
msg_merge.head(2)
```

```python
msg_merge[msg_merge['저니'].notna()].to_csv('./data/MSG.csv')
```

```python
cust_list = list(set(msg_merge['INCS_NO']))
cust_df = pd.DataFrame(cust_list, columns=['INCS_NO'])
print(cust_df.shape)
cust_df.head(2)
```

```python
# log 1,2,3 모두 처리
i = 2
input_path = f'./raw/LOG{i}.csv'
output_path = f'./data/LOG{i}.csv'

log = pd.read_csv(input_path)
print(log.shape)

log_sampl = pd.merge(log, cust_df, on='INCS_NO', how='inner')
print(log_sampl.shape)

log_sampl.to_csv(output_path)
```
