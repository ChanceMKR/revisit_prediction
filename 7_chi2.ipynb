```python
import pandas as pd
```

```python
control = pd.read_csv('./data/control_sample.csv', index_col=0)
print(control.shape)
```

```python
sm1 = pd.read_csv('./data/sample1.csv', index_col=0)
print(sm1.shape)
sm1.head(2)
```

```python
sm2 = pd.read_csv('./data/sample2.csv', index_col=0)
print(sm2.shape)
sm2.head(2)
```

```python
treat = pd.concat([sm1, sm2])
print(treat.shape)
treat.head(2)
```

```python
control_counts = control['y'].value_counts()
treat_counts = treat['y'].value_counts()

# 빈도표 결합
contingency_table = pd.DataFrame({'control': control_counts, 'treat': treat_counts}).fillna(0)
print(contingency_table)
```

```python
from scipy.stats import chi2_contingency

# chi2 검정 수행
chi2, p, dof, expected = chi2_contingency(contingency_table)
print('Chi2 통계량:', chi2)
print('p-value:', p)
```
